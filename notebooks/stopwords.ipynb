```python
## speech of dr apj abdul kalam

paragraph = "I am delighted to be here with you today. I am grateful to the organizers for giving me this opportunity to share my thoughts with you. I am also thankful to all of you for taking the time to listen to me. I hope that my words will inspire you and motivate you to achieve your dreams. Remember, the future belongs to those who believe in the beauty of their dreams. So, dream big and work hard to make your dreams come true. Thank you."
```

```python
from nltk.stem import PorterStemmer
```

```python
from nltk.corpus import stopwords
```

```python
import ssl
import certifi

ssl._create_default_https_context = lambda: ssl.create_default_context(cafile=certifi.where())

import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
stopwords.words('french')
```

```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()
```

```python
sentences = nltk.sent_tokenize(paragraph)
```

```python
type(sentences)
```

```python
### Apply stopwords and filter and then apply stemming to the sentences

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words= [stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words) # converting all list of words into sentences
```

```python
sentences 
```

```python
from nltk.stem import SnowballStemmer
snowballstemmer = SnowballStemmer('english')
```

```python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words= [snowballstemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words) # converting all list of words into sentences
```

```python
sentences 
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words= [lemmatizer.lemmatize(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words) # converting all list of words into sentences
```
